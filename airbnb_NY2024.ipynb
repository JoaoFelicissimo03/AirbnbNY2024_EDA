### **5.2 Feature Engineering**

```python
# Custom color palette for neighbourhood groups
nyc_palette = {
    "Manhattan": "#8472EB",
    "Brooklyn": "#887067",
    "Queens": "#C21818C8",
    "Staten Island": "#00791A",
    "Bronx": "#53024D",
}
```

```python
# Plotting the average price by neighbourhood group

df.groupby("neighbourhood_group")["price"].mean().sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 6),
    color=[
        nyc_palette[group]
        for group in df.groupby("neighbourhood_group")["price"]
        .mean()
        .sort_values(ascending=False)
        .index
    ],
)
plt.ylabel("Average Price ($)")
plt.title("Average Price by Neighbourhood Group in New York City")
```

```python
# Create a new column 'price_per_bed' by dividing the 'price' column by the 'beds' column
df["price_per_bed"] = df["price"] / df["beds"]

df.groupby("neighbourhood_group")["price_per_bed"].mean()
```

```python
sns.barplot(
    x="neighbourhood_group",
    y="price_per_bed",
    data=df,
    estimator=np.mean,
    hue="neighbourhood_group",
    palette=nyc_palette,
)
plt.ylabel("Average Price per Bed ($)")
plt.title("Average Price per Bed by Neighbourhood Group in New York City")
```

```python
# Function/Logic to categorize the license
def cat_license(val):
    if pd.isna(val) or val == "No License":
        return "Unlicensed"
    elif val == "Exempt":
        return "Exempt"
    elif str(val).startswith("OSE-STRREG"):
        return "Licensed"
    else:
        return "Licensed" if "STRREG" in str(val) else "Unlicensed"


# Applying the categorization function to create a new column 'license_category'
df["license_category"] = df["license"].apply(cat_license)

# Color palette for license categories
license_colors = {"Unlicensed": "#FF5A5F", "Licensed": "#00A699", "Exempt": "#357DED"}
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(
    x="license_category", data=df, hue="license_category", palette=license_colors
)
plt.title("Count of Airbnb Listings by License Category in New York City")
plt.xlabel("License Category")
plt.ylabel("Count")
```

### **5.3 Bi Variate Analysis**


```python
#   Custom color palette for room types
room_palette = {
    "Private room": "#00A699",
    "Entire home/apt": "#FF5A5F",
    "Hotel room": "#FC642D",
    "Shared room": "#484848",
}
```

```python
# Plotting the average price per bed by neighbourhood group and room type

sns.barplot(
    x="neighbourhood_group",
    y="price_per_bed",
    data=df,
    hue="room_type",
    estimator=np.mean,
    palette=room_palette,
)
plt.ylabel("Average Price per Bed ($)")
plt.title("Average Price per Bed by Neighbourhood Group and Room Type in New York City")
```

```python
# Scatter plot of number of reviews vs price, colored by neighbourhood group

plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="number_of_reviews",
    y="price",
    data=df,
    hue="neighbourhood_group",
    palette=nyc_palette,
    alpha=0.7,
)
plt.ylabel("Price ($)")
plt.title("Number of Reviews vs Price by Neighbourhood Group in New York City")
```

```python
plt.figure(figsize=(12, 6))

ax = sns.countplot(
    data=df,
    x="neighbourhood_group",
    hue="license_category",
    palette=license_colors,
    order=["Manhattan", "Brooklyn", "Queens", "Bronx", "Staten Island"],
)

plt.title(
    "NYC Airbnb Listings: License Category by Neighborhood Group",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Neighborhood Group", fontsize=12)
plt.ylabel("Number of Listings", fontsize=12)
plt.legend(title="License Status")

plt.tight_layout()
plt.show()
```

```python
# Pairplot of selected numerical features, colored by room type
sns.pairplot(
    data=df,
    vars=[
        "price",
        "minimum_nights",
        "number_of_reviews",
        "availability_365",
        "price_per_bed",
    ],
    hue="room_type",
)
```

```python
# Correlation matrix of numerical features
filtered_columns = [
    "latitude",
    "longitude",
    "price",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "availability_365",
    "beds",
]
correlation_matrix = df[filtered_columns].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Numerical Features in New York City AirBNB Dataset")
```

### **5.4 Geographic Analysis**

```python
# Importing contextilt and folium for geographical visualization
import contextily as ctx
import folium
from folium.plugins import MarkerCluster
```

```python
df.columns
```

```python
df["neighbourhood_group"].value_counts()
```

```python
plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    x="longitude",
    y="latitude",
    data=df,
    hue="neighbourhood_group",
    alpha=0.7,
    palette=nyc_palette,
)

ctx.add_basemap(ax, crs="EPSG:4326", source=ctx.providers.CartoDB.Positron)

plt.title("Geographical Distribution of Airbnb Listings in NYC")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    x="longitude",
    y="latitude",
    data=df,
    hue="room_type",
    alpha=0.7,
    palette=room_palette,
)

ctx.add_basemap(ax, crs="EPSG:4326", source=ctx.providers.CartoDB.Positron)

plt.title("Geographical Distribution of Airbnb Listings in NYC by Room Type")
plt.show()
```

```python
plt.figure(figsize=(12, 8))

ax = sns.scatterplot(
    x="longitude",
    y="latitude",
    data=df,
    hue="license_category",
    palette=license_colors,
    alpha=0.6,
    s=15,
)

ctx.add_basemap(ax, crs="EPSG:4326", source=ctx.providers.CartoDB.Positron)

plt.title(
    "Geographical Distribution of NYC Airbnb Listings by License Status (Local Law 18)"
)
plt.legend(title="License Status", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()

plt.show()
```

**5.4.1 HTML Cluster Map Creation**

```python
# 1. Create the base map centered in New York City
map = folium.Map(location=[40.7128, -74.0060], zoom_start=11, tiles="CartoDB Positron")

# 2. Create the cluster (grouping of markers)
marker_cluster = MarkerCluster().add_to(map)

# 3. Add each listing from the DataFrame to the cluster
for idx, row in df.dropna(subset=["latitude", "longitude"]).iterrows():
    folium.Marker(
        location=[row["latitude"], row["longitude"]],
        popup=f"Neighborhood: {row.get('neighbourhood', 'N/A')}",
    ).add_to(marker_cluster)

# 4. Save the map to an HTML file to open in the browser
map.save("cluster_airbnb_map.html")
print("Map saved! Open the file 'cluster_airbnb_map.html' in your browser.")
```
