### **5.2 Feature Engineering**

```python
# Custom color palette for neighbourhood groups
nyc_palette = {
    "Manhattan": "#8472EB",
    "Brooklyn": "#887067",
    "Queens": "#C21818C8",
    "Staten Island": "#00791A",
    "Bronx": "#53024D",
}
```

```python
# Plotting the average price by neighbourhood group

df.groupby("neighbourhood_group")["price"].mean().sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 6),
    color=[
        nyc_palette[group]
        for group in df.groupby("neighbourhood_group")["price"]
        .mean()
        .sort_values(ascending=False)
        .index
    ],
)
plt.ylabel("Average Price ($)")
plt.title("Average Price by Neighbourhood Group in New York City")
```

```python
# Create a new column 'price_per_bed' by dividing the 'price' column by the 'beds' column
df["price_per_bed"] = df["price"] / df["beds"]

df.groupby("neighbourhood_group")["price_per_bed"].mean()
```

```python
sns.barplot(
    x="neighbourhood_group",
    y="price_per_bed",
    data=df,
    estimator=np.mean,
    hue="neighbourhood_group",
    palette=nyc_palette,
)
plt.ylabel("Average Price per Bed ($)")
plt.title("Average Price per Bed by Neighbourhood Group in New York City")
```

```python
# Function/Logic to categorize the license
def cat_license(val):
    if pd.isna(val) or val == "No License":
        return "Unlicensed"
    elif val == "Exempt":
        return "Exempt"
    elif str(val).startswith("OSE-STRREG"):
        return "Licensed"
    else:
        return "Licensed" if "STRREG" in str(val) else "Unlicensed"


# Applying the categorization function to create a new column 'license_category'
df["license_category"] = df["license"].apply(cat_license)

# Color palette for license categories
license_colors = {"Unlicensed": "#FF5A5F", "Licensed": "#00A699", "Exempt": "#357DED"}

plt.figure(figsize=(10, 6))
sns.countplot(
    x="license_category", data=df, hue="license_category", palette=license_colors
)
plt.title("Count of Airbnb Listings by License Category in New York City")
plt.xlabel("License Category")
plt.ylabel("Count")
```

```python
# 1. Cleaning the 'rating' column to handle non-numeric values and missing data
def clean_rating(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()
    if val_str in ["new", "no ratings", "none", "", "nan"]:
        return np.nan

    try:
        # Remove any non-numeric characters and convert to float
        val_clean = val_str.replace("/5", "").replace("stars", "").strip()
        num = float(val_clean)
        # Garante que o rating está dentro de uma escala válida (ex: 0 a 5)
        return num if 0 <= num <= 5 else np.nan
    except ValueError:
        return np.nan


# Apply the cleaning function to the 'rating' column and create a new column 'rating_clean'
df["rating_clean"] = df["rating"].apply(clean_rating)

# Verifying the cleaning process by checking the number of valid and invalid ratings
print("After cleaning the 'rating' column:")
print(f"Valid numeric values: {df['rating_clean'].notna().sum()}")
print(f"Listings without rating (New/Empty): {df['rating_clean'].isna().sum()}")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["rating_clean"].dropna(), edgecolor="black")
plt.title("Distribution of Ratings for Airbnb Listings in New York City")
plt.xlabel("Cleaned Rating")
plt.ylabel("Frequency")
```

### **5.4 Geospacial Analysis**

```python
# Importing contextilt and folium for geographical visualization
import contextily as ctx
import folium
from folium.plugins import MarkerCluster
import geopandas as gpd
```

**5.4.1 Creating Context Map(s)**

```python
# Load official NYC Borough Boundaries (GeoJSON from NYC Open Data / GitHub)
url = "https://raw.githubusercontent.com/dwillis/nyc-maps/master/boroughs.geojson"
nyc_geojson = gpd.read_file(url)

# Ensure CRS is EPSG:4326 for web map compatibility
nyc_geojson = nyc_geojson.to_crs(epsg=4326)

# Map the polygon column names (usually BoroName) to your palette keys
# (Adjusting column name if needed: nyc_geojson['BoroName'])
nyc_geojson["color"] = nyc_geojson["BoroName"].map(nyc_palette)

# Create the Context Map
fig, ax = plt.subplots(figsize=(12, 10))

# Plot the borough boundaries with filled colors (Choropleth/Context Style)
nyc_geojson.plot(
    ax=ax,
    color=nyc_geojson["color"],
    edgecolor="white",  # Clean white borders between boroughs
    linewidth=1.2,
    alpha=0.6,  # Transparency to show base map context underneath
)

# Add basemap tile for spatial context (water bodies, roads, surrounding regions)
ctx.add_basemap(ax, crs="EPSG:4326", source=ctx.providers.CartoDB.Positron)

# Add Borough Labels directly on the map for clear spatial context
for idx, row in nyc_geojson.iterrows():
    # Find the central coordinate of each borough polygon
    centroid = row.geometry.centroid
    ax.annotate(
        text=row["BoroName"],
        xy=(centroid.x, centroid.y),
        horizontalalignment="center",
        verticalalignment="center",
        fontsize=11,
        fontweight="bold",
        color="black",
        bbox=dict(
            boxstyle="round,pad=0.3", fc="white", ec="none", alpha=0.7
        ),  # Clean label badge
    )

# 5. Titles and Formatting (All in English)
plt.title(
    "Study Area Context Map: NYC Neighborhood Groups (Boroughs)",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Longitude", fontsize=11)
plt.ylabel("Latitude", fontsize=11)

# Clean grid lines
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()

plt.show()
```

**5.4.3 HTML Cluster Map Creation**

```python
# 1. Create the base map centered in New York City
map = folium.Map(location=[40.7128, -74.0060], zoom_start=11, tiles="CartoDB Positron")

# 2. Create the cluster (grouping of markers)
marker_cluster = MarkerCluster().add_to(map)

# 3. Add each listing from the DataFrame to the cluster
for idx, row in df.dropna(subset=["latitude", "longitude"]).iterrows():
    folium.Marker(
        location=[row["latitude"], row["longitude"]],
        popup=f"Neighborhood: {row.get('neighbourhood', 'N/A')}",
    ).add_to(marker_cluster)

# 4. Save the map to an HTML file to open in the browser
map.save("cluster_airbnb_map.html")
print("Map saved! Open the file 'cluster_airbnb_map.html' in your browser.")
```
