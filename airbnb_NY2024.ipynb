### **4. Data Cleaning**

**4.1 Missing data**

```python
data.isnull().sum()
```

```python
## Dropping rows with missing values
data.dropna(inplace=True)

data.isnull().sum()
```

**4.2 Duplicate data**

```python
data.duplicated().sum()
```

```python
## Dropping duplicate rows
data.drop_duplicates(inplace=True)
data.duplicated().sum() 
```

**4.3 Data types**

```python
data.dtypes
```

```python
#Changing the data types of 'id' and 'host_id' columns to object type
data['id'] = data['id'].astype(object)
data['host_id'] = data['host_id'].astype(object)

data.dtypes
```

**4.4 Saving Clean Data**

```python
data.to_csv('data/processed/airbnb_dataset_clean.csv', index=False)
```

```python
## Checking the shape of the raw and cleaned datasets to ensure that the cleaning process has been successful
dataraw = pd.read_csv('data/raw/airbnb_dataset.csv', encoding_errors='ignore')
dataclean = pd.read_csv('data/processed/airbnb_dataset_clean.csv', encoding_errors='ignore')

dataraw.shape, dataclean.shape
```

## **5. Data Analysis**

### **5.1 Univariate Analysis**

**5.1.1 Univariate Analysis - Price**

```python
## Outlier detection

sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
sns.boxplot(x=data['price'])
```

```python
df = data[data['price'] < 1500]

sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
sns.boxplot(x=df['price'])
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
sns.histplot(df['price'], kde=True, bins=50)
plt.ylabel('Frequency')
plt.title('Price Distribution of Airbnb Listings in New York City')
```

**5.1.2 Univariate Analysis - Availability**

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
sns.histplot(df['availability_365'], kde=True)
plt.ylabel('Frequency')
plt.title('Availability Distribution of Airbnb Listings in New York City')
```

### **5.2 Feature Engineering**

```python
# Plotting the average price by neighbourhood group

df.groupby('neighbourhood_group')['price'].mean().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
plt.ylabel('Average Price')
plt.title('Average Price by Neighbourhood Group in New York City')
```

```python
# Create a new column 'price_per_bed' by dividing the 'price' column by the 'beds' column
df['price_per_bed']= df['price']/df['beds']
df.head(1)
```

```python
df.groupby('neighbourhood_group')['price_per_bed'].mean().sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
sns.set_theme(style="darkgrid", palette="deep", font='sans-serif', font_scale=1.1)
plt.ylabel('Average Price per Bed')
plt.title('Average Price per Bed by Neighbourhood Group in New York City')
```
